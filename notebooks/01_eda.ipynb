# 1. EDA - Разведочный анализ данных

## Датасет
- **Классы:** real (0), fake (1)
- **Train:** 50 000 изображений, **Test:** 10 000 изображений
- **Метки:** `dataset/train_solution.csv` (idx_image, real/fake)
- **Особенности:** дисбаланс классов (41 500 real vs 8 500 fake), наличие шума

```python
import sys
sys.path.append("..")

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image

from src.config import CSV_PATH, IMAGES_DIR, FIGURES_DIR, SEED
from src.utils import set_seed

set_seed(SEED)
%matplotlib inline

FIGURES_DIR.mkdir(parents=True, exist_ok=True)
```

```python
df = pd.read_csv(CSV_PATH, header=None, names=["idx_image", "real/fake"])
print(f"Размер датасета: {df.shape}")
df.head()
```

## Баланс классов

```python
counts = df["real/fake"].value_counts().sort_index()
print(counts)
print(f"\nДисбаланс: {counts[0] / counts[1]:.2f}  (real : fake)")

plt.figure(figsize=(6, 4))
sns.barplot(x=["real", "fake"], y=counts.values)
plt.title("Распределение классов")
plt.ylabel("Количество изображений")
plt.savefig(FIGURES_DIR / "class_balance.png", dpi=150, bbox_inches="tight")
plt.show()
```

## Примеры изображений

```python
fig, axes = plt.subplots(2, 5, figsize=(15, 6))
for i, ax in enumerate(axes.flat):
    row = df.sample(1, random_state=SEED + i).iloc[0]
    img_path = IMAGES_DIR / f"{row['idx_image']}.jpg"
    ax.imshow(Image.open(img_path))
    ax.set_title(f"{'fake' if row['real/fake'] == 1 else 'real'}")
    ax.axis("off")
plt.tight_layout()
plt.savefig(FIGURES_DIR / "sample_images.png", dpi=150, bbox_inches="tight")
plt.show()
```
